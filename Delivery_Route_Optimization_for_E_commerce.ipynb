Profiling (Time & Memory)

```python
from itertools import permutations

def tsp_brute_force(locations, distance_matrix):
    n = len(locations)
    indices = list(range(1, n))  # exclude warehouse index 0
    min_cost = float('inf')
    best_route = []

    for perm in permutations(indices):
        cost = distance_matrix[0][perm[0]]
        for i in range(len(perm) - 1):
            cost += distance_matrix[perm[i]][perm[i+1]]
        cost += distance_matrix[perm[-1]][0]  # return to warehouse

        if cost < min_cost:
            min_cost = cost
            best_route = perm

    return [locations[0]] + [locations[i] for i in best_route] + [locations[0]], min_cost
```

```python
import time

sizes = [3, 4, 5, 6]   # number of locations including Warehouse
times = []

for n in sizes:
    # Take first n locations
    loc_n = locations[:n]

    # Slice the distance matrix to n x n
    dist_n = [row[:n] for row in distance_matrix[:n]]

    start = time.perf_counter()
    route, cost = tsp_brute_force(loc_n, dist_n)
    end = time.perf_counter()

    elapsed = end - start
    times.append(elapsed)

    print(f"n = {n} locations → time = {elapsed:.6f} s, route = {route}, cost = {cost}")
```

```python
plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("Number of Locations")
plt.ylabel("Time (seconds)")
plt.title("TSP Brute Force Time vs Locations")
plt.grid(True)
plt.show()
```
